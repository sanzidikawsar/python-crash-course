## Domain crib sheet (you know code; here's the science the terms mean)
You are a CS grad, not a biochemist — you never need the science, only enough to answer “what is that?” without stalling. The plan is built so you teach the *shape*, and students supply the domain.

**Session data (Parts 2–3):**
- `absorbance` — a light-absorption reading from a spectrophotometer; higher = more of the substance present. Just a measurement column.
- `ph` — acidity, 0–14 (7 = neutral). A number.
- `treatment` = `Control` vs `Treated` — experimental groups (control got nothing, treated got the intervention). Think A/B test.
- `replicate` — the same experiment repeated (1/2/3) for reliability.
- `n.d.` — “not detected” (below the instrument's limit); this is why `absorbance` loads as text.

**Part 4 — Genomics (Biopython):**
- A DNA **sequence** is a string of letters `A/T/G/C` (the bases).
- **GC content** = the percentage of bases that are `G` or `C`. A standard, simple property (higher GC ≈ more thermally stable). Literally count G+C, divide by length.
- **Biopython** (`Bio`) — the standard Python library for biological sequences (reading FASTA files, computing sequence properties).
- **FASTA** — the standard text format for sequences: a `>` header line, then the letters. `SeqIO.parse` loops over its records.
- **Central dogma**: DNA → (transcribe) → RNA → (translate) → protein. `.transcribe()` swaps T for U; `.translate()` reads codons (letter triples) into amino acids. That protein string is the real gene product.
- **Reverse complement** — the partner DNA strand (A↔T, G↔C, reversed); `.reverse_complement()`.
- **Monocots vs dicots** — two big plant groups (rice/maize/sorghum vs Arabidopsis/soybean/tomato). Monocot genes really do run GC-richer — the bar chart shows it.

**Part 4 — Molecular dynamics (MDAnalysis):**
- **Molecular dynamics (MD)** — a physics simulation of how a molecule (here a protein) wiggles over time. Output is a **trajectory**: many snapshots (**frames**), each holding the 3-D positions of every atom.
- **Radius of gyration** — one number for how compact vs spread-out the protein is in a frame; it moves as the protein folds/unfolds. We compute one per frame → a table over time.
- **MDAnalysis** (`mda`) — the standard Python library to read MD trajectories and measure things. **PSF/DCD** are file formats (PSF = which atoms exist, DCD = their positions per frame); the sample files ship with the library.
- **Residue** — one amino-acid unit of the protein chain; the sample has 214.
- **RMSD** (root-mean-square deviation) — one number for how far the protein moved from the first frame; the classic MD “how much did it change?” metric.

**Why the two tracks are one lesson:** open → loop over records → one number each → collect into a table → summarise/plot. Absorbance-per-file, GC-per-sequence, and radius-per-frame are the same shape. Teach the shape.

```python
# RUN THIS (load both field tools) — just run it, no typing:
import warnings
warnings.filterwarnings("ignore")
import urllib.request
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
import MDAnalysis as mda
from MDAnalysis.tests.datafiles import PSF, DCD
print("field tools loaded")
```

## Genomics · Biopython

Sequences live in **FASTA** files. We download a small file of plant genes and loop over its records — *open → loop → one number each*, exactly like looping over the twelve CSV files earlier:

```python
# RUN THIS (genomics — read the FASTA, GC% of each gene) — just run it, no typing:
url = "https://raw.githubusercontent.com/sanzidikawsar/python-crash-course/main/data/genomics/plant_genes.fasta"
urllib.request.urlretrieve(url, "plant_genes.fasta")
for gene in SeqIO.parse("plant_genes.fasta", "fasta"):
    print(gene.id, "| length:", len(gene.seq), "| GC%:", round(gc_fraction(gene.seq) * 100, 1))
```

**The central dogma, in three lines.** A gene is DNA; it is transcribed to RNA, then translated to a protein. Biopython does each step with one call — your gene becomes the actual protein letters:

```python
# RUN THIS (genomics — turn a gene into its protein) — just run it, no typing:
first_gene = next(SeqIO.parse("plant_genes.fasta", "fasta"))
print("DNA    :", first_gene.seq[:30], "...")
print("RNA    :", first_gene.seq.transcribe()[:30], "...")
print("protein:", first_gene.seq.translate(to_stop=True))
```

**The other strand.** DNA is double-stranded; the *reverse complement* is the partner strand, read the other way:

```python
# RUN THIS (genomics — the complementary strand) — just run it, no typing:
piece = first_gene.seq[:12]
print("strand 1:", piece)
print("strand 2:", piece.reverse_complement())
```

**One picture.** Collect every gene's GC% and bar-chart it — you can *see* that monocots (rice, maize, sorghum) run GC-richer than dicots (Arabidopsis, soybean, tomato). Same plotting as Part 3:

```python
# RUN THIS (genomics — bar chart of GC% per gene) — just run it, no typing:
ids = []
gc = []
for gene in SeqIO.parse("plant_genes.fasta", "fasta"):
    ids.append(gene.id)
    gc.append(round(gc_fraction(gene.seq) * 100, 1))
plt.bar(ids, gc)
plt.title("GC % per plant gene")
plt.show()
```

## Molecular dynamics · MDAnalysis

A molecular-dynamics run is a movie of a protein wobbling: many **frames**, each holding the 3-D positions of every atom. First, what is inside it:

```python
# RUN THIS (MD — what is in the simulation) — just run it, no typing:
warnings.filterwarnings("ignore")
universe = mda.Universe(PSF, DCD)
print("atoms:", len(universe.atoms))
print("residues:", len(universe.residues))
print("frames:", len(universe.trajectory))
```

**One number per frame.** The *radius of gyration* measures how compact the protein is. We loop over frames and collect it — the same *open → loop → one number each* shape as GC-per-gene:

```python
# RUN THIS (MD — protein size per frame) — just run it, no typing:
protein = universe.select_atoms("protein")
sizes = []
for frame in universe.trajectory:
    sizes.append(round(float(protein.radius_of_gyration()), 2))
print("first 5 of", len(sizes), "frames:", sizes[:5])
```

**One picture.** Plot the protein's size across the whole simulation:

```python
# RUN THIS (MD — plot protein size over time) — just run it, no typing:
plt.plot(sizes, marker="o")
plt.xlabel("Frame")
plt.ylabel("Radius of gyration")
plt.title("Protein size over the simulation")
plt.show()
```

**How far did it move?** RMSD compares every frame back to the first and reports, in one number, how much the protein moved during the run:

```python
# RUN THIS (MD — total movement (RMSD)) — just run it, no typing:
from MDAnalysis.analysis import rms
result = rms.RMSD(universe, select="protein").run()
moved = result.results.rmsd[:, 2]
print("protein moved", round(float(moved[-1]), 2), "Angstrom start to end")
```
